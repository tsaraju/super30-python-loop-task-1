```python
#Write a program to print numbers from 1 to 100 using a for loop
for i in range(1, 101):
    print(i)
```

```python
#Print all even numbers from 1 to 100
for i in range(1, 101):
    if i % 2 == 0: print(i) 
```

```python
#Print all odd numbers from 1 to 100
for i in range(1, 101):
    if i % 2 != 0: print(i) 
```

```python
#Take an integer n and print its multiplication table from 1 to 20
# Input 10
n = int(input("Enter a number: "))
for i in range(1, 21):
    print(f'{n} * {i} = {n*i}')
```

```python
#Calculate the sum of numbers from 1 to n using a loop
#Input 10
n = int(input("Enter a number: "))
sum = 0
for i in range(1, n+1):
    sum = sum + i
print(f'The sum of numbers from 1 to {n} is: {sum}')
```

```python
#Calculate the factorial of a number without using any built-in factorial function
# Input 5
n = int(input("Enter a number: "))
fact = 1
for i in range(0, n+1):
    if i == 0 or i == 1:
        fact = 1
    else:
        fact = fact * i

print(f'The factorial of the number {n} is: {fact}')
```

```python
#Print only the numbers divisible by 3
numbers = [12, 7, 9, 20, 33, 42, 8, 15]
for i in numbers:
    if i % 3 == 0:
        print(f'{i} is divisible by 3')
```

```python
#Print every language along with its length
languages = ["Python", "Java", "C++", "JavaScript", "Go"]
for l in languages:
    print(f'{l} language and its length is: {len(l)}')
```

```python
#Print every key and value
student = {
"name": "Rahul",
"age": 22,
"course": "Data Science",
"city": "Bangalore"
}

for key, value in student.items():
    print(f'key {key}: value {value}')
```

```python
#Count how many vowels exist in a user-provided string
# Input abe
str = input("Enter a string: ")
cnt = 0
for i in str:
    if i in ['a','e','i','o','u']:
        cnt += 1

print(f'The number of vowels in the string is: {cnt}')
```

```python
#Reverse a string using a for loop without using [::-1] or reversed()
s = input("Enter a string: ")

for i in range(len(s)-1, -1, -1):
    print(s[i])
```

```python
#Find the largest number from a list without using max()
list1 = [21, 2, 13, 64, 45]
large = list1[0]
for i in list1:
    if i > large:
        large = i

print(f'The largest number in the list is: {large}')
```
